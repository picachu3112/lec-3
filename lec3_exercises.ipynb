````python
import math

def area_of_circle(r):
    area = math.pi * r * r
    return area

def perimeter_of_circle(r):
    perimeter = 2 * math.pi * r
    return perimeter
````

````python
def add_all_nums(*args):
    for num in args:
        if not isinstance(num, (int, float)):
            return "All arguments must be numbers."
    
    return sum(args)
````

````python
def check_season(month):
    if month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    elif month in ["June", "July", "August"]:
        return "Summer"
    elif month in ["September", "October", "November"]:
        return "Autumn"
    else:
        return "Invalid month"
````

````python
check_season("July")
````

````python
def calculate_slope(x1, y1, x2, y2):
    slope = (y2 - y1) / (x2 - x1)
    return slope
````

````python
calculate_slope(1, 2 ,3, 6)
````

````python
import math

def solve_quadratic_eqn(a, b, c):
    discriminant = b**2 - 4*a*c
    
    x1 = (-b + math.sqrt(discriminant)) / (2*a)
    x2 = (-b - math.sqrt(discriminant)) / (2*a)
    
    return x1, x2
````

````python
solve_quadratic_eqn(1, -5, 6)
````

````python
def sum_of_odds(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total
````

````python
def sum_of_even(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i

    return total
````

````python
def reverse_list(items):
    reversed_list = []

    for i in range(len(items) - 1, -1, -1):
        reversed_list.append(items[i])

    return reversed_list
````

````python
print(reverse_list([1, 2, 3, 4, 5]))
print(reverse_list(["A", "B", "C" ,"D"]))
````

````python
def evens_and_odds(n):
    evens = 0
    odds = 0

    for i in range(n + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1

    print(f"The number of odds are {odds}.")
    print(f"The number of evens are {evens}.")
````

````python
def calculate_mean(numbers):
    total = sum(numbers)
    count = len(numbers)
    
    mean = total / count
    
    return mean

def calculate_median(numbers):
    numbers = sorted(numbers)
    n = len(numbers)

    if n % 2 != 0:
        median = numbers[n // 2]
    else:
        middle1 = numbers[n // 2 - 1]
        middle2 = numbers[n // 2]
        median = (middle1 + middle2) / 2

    return median
def calculate_mode(numbers):
    frequency = {}

    for num in numbers:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1

    mode = max(frequency, key=frequency.get)

    return mode
def calculate_mode(numbers):
    max_count = 0
    mode = None

    for num in numbers:
        count = 0

        for x in numbers:
            if x == num:
                count += 1

        if count > max_count:
            max_count = count
            mode = num

    return mode
````

````python
numbers = [2, 5, 3, 2, 8, 5, 2, 7, 5, 2, 9]
print(calculate_mean(numbers))
print(calculate_median(numbers))
print(calculate_mode(numbers))
````

````python
def is_same_type(items):
    for item in items:
        if type(item) != type(items[0]):
            return False

    return True
````

````python
print(is_same_type([1, 2, 3, 4]))
print(is_same_type(["a", "b", "c"]))
print(is_same_type([1, "2", 3]))
````

````python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
````

````python
print(is_prime(7))
print(is_prime(10))
print(is_prime(13))
print(is_prime(1))
````

````python
def are_unique(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return False

    return True
````

````python
print(are_unique([1, 2, 3, 4]))
print(are_unique([1, 2, 3, 2]))
print(are_unique(["a", "b", "c"]))
print(are_unique(["a", "b", "a"]))
````
