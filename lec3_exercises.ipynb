````python
import math

def area_of_circle(r):
    area = math.pi * r * r
    return area

def perimeter_of_circle(r):
    perimeter = 2 * math.pi * r
    return perimeter
````

````python
print(area_of_circle(10))
print(perimeter_of_circle(20))
````

````python
def add_all_nums(*args):
    for num in args:
        if not isinstance(num, (int, float)):
            return "All arguments must be numbers."
    
    return sum(args)
````

````python
add_all_nums(5,6,5,3,2,4,6)
````

````python
def celsius_2_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit
````

````python
celsius_2_fahrenheit(37)
````

````python
def check_season(month):
    if month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    elif month in ["June", "July", "August"]:
        return "Summer"
    elif month in ["September", "October", "November"]:
        return "Autumn"
    else:
        return "Invalid month"
````

````python
check_season("July")
````

````python
def calculate_slope(x1, y1, x2, y2):
    slope = (y2 - y1) / (x2 - x1)
    return slope
````

````python
calculate_slope(1, 2 ,3, 6)
````

````python
import math

def solve_quadratic_eqn(a, b, c):
    discriminant = b**2 - 4*a*c
    
    x1 = (-b + math.sqrt(discriminant)) / (2*a)
    x2 = (-b - math.sqrt(discriminant)) / (2*a)
    
    return x1, x2
````

````python
solve_quadratic_eqn(1, -5, 6)
````

````python
def print_list(items):
    for item in items:
        print(item)
````

````python
print_list(["Python", "Java", "C++"])
````

````python
def capitalize_list_items(items):
    capitalized_items = []

    for item in items:
        capitalized_items.append(item.capitalize())

    return capitalized_items
````

````python
capitalize_list_items(["apple", "banana", "mango"])
````

````python
def add_item(items, item):
    items.append(item)
    return items
````

````python
numbers = [2, 3, 7, 9]
print(add_item(numbers, 5))
````

````python
def remove_item(items, item):
    items.remove(item)
    return items
````

````python
numbers = [2, 3, 7, 9, 5, 6,8]
print(remove_item(numbers, 5))
````

````python
def sum_of_numbers(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total
````

````python
print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(3))
````

````python
def sum_of_odds(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total
````

````python
sum_of_odds(5)
````

````python
def sum_of_even(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i

    return total
````

````python
sum_of_even(5)
````
